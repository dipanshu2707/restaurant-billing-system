```python
     







menu = {
    # Starters
    'Veg Spring Roll': 120,
    'Paneer Tikka': 180,
    'Chicken Wings': 200,

    # Main Course
    'Butter Paneer Masala': 250,
    'Dal Tadka': 180,
    'Chicken Biryani': 300,
    'Veg Fried Rice': 200,

    # Breads
    'Tandoori Roti': 25,
    'Garlic Naan': 40,
    'Butter Naan': 50,

    # Desserts
    'Gulab Jamun (2 pcs)': 80,
    'Ice Cream (Scoop)': 100,
    'Chocolate Brownie': 150,

    # Beverages
    'Masala Chai': 40,
    'Cold Coffee': 120,
    'Fresh Lime Soda': 90,
}

print("Welcome to Rajput Restaurant")
print("======================================")

print("\nMenu:")
for item, price in menu.items():
    print(f"{item} : ₹{price}")

order_total = 0

while True:
    item = input("\nEnter the item you want to order = ")

    # Check item ignoring capital/small letters
    found_item = None

    for menu_item in menu:
        if menu_item.lower() == item.lower():
            found_item = menu_item
            break

    if found_item:
        order_total += menu[found_item]
        print(f"{found_item} has been added to your order.")
        print(f"Price: ₹{menu[found_item]}")
    else:
        print(f"Sorry, {item} is not available.")

    another_order = input("\nDo you want to add another item? (yes/no) = ")

    if another_order.lower() != "yes" and another_order.lower() != "y":
        break

print("\n======================================")
print(f"Total bill = ₹{order_total}")
print("Thank you for visiting Rajput Restaurant!")


```
